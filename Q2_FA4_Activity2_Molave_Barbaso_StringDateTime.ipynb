```python
# Version 1: Basic input cleaning and output (No date formatting yet)

event_title = "  project submission deadline  "
year = 2026
month = 11
day = 25

# Clean string input
formatted_event = event_title.strip().title()

print("--- Basic Output ---")
print(f"Event: {formatted_event}")
print(f"Date: {year}-{month}-{day}")
print(f"Year: {year}")
print(f"Month: {month}")
print(f"Day: {day}")
```

```python
# Version 2: Added datetime module and readable date formatting
from datetime import datetime

event_title = "  project submission deadline  "
year = 2026
month = 11
day = 25

# Create datetime object and clean event title
event_date = datetime(year, month, day)
formatted_event = event_title.strip().title()

print("--- Formatted Output ---")
print(f"Event: {formatted_event}")
print(f"Date: {event_date.strftime('%B %d, %Y')}")
print(f"Year: {event_date.year}")
print(f"Month: {event_date.strftime('%B')}")
print(f"Day: {event_date.day}")
print(f"Weekday: {event_date.strftime('%A')}")
```
