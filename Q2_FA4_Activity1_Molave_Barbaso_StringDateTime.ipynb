```python
# Version 2: Added datetime module and readable deadline formatting
from datetime import datetime

first_name = "  juan "
last_name = " dela cruz "
section = "narra "
fav_subject = "computer science"

# Cleaning and formatting logic
first_name = first_name.strip().title()
last_name = last_name.strip().title()
section = section.strip().title()
fav_subject = fav_subject.strip().title()

full_name = first_name + " " + last_name
full_name_upper = full_name.upper()
formatted_section = section
formatted_subject = fav_subject
char_count = len(full_name)

# Date and Time features
current_time = datetime.now()
formatted_submitted = current_time.strftime("%B %d, %Y at %I:%M %p")

deadline = datetime(2026, 11, 25, 17, 0)  # Year, Month, Day, Hour, Minute
formatted_deadline = deadline.strftime("%A, %B %d, %Y at %I:%M %p")

print("--- Formatted Output ---")
print(f"Name: {full_name}")
print(f"Name Code: {full_name_upper}")
print(f"Section: {formatted_section}")
print(f"Favorite Subject: {formatted_subject}")
print(f"Character Count: {char_count}")
print(f"Date Submitted: {formatted_submitted}")
print(f"Project Deadline: {formatted_deadline}")
```
